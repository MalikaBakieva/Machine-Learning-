```python
import numpy as np

x=np.array([3,5,4,7])
y=np.array([5,2,4,9])

def cost(w):
  pred = w*x
  return np.mean((pred-y)**2)

for w in [2,3,4,5]:
  print(f"w={w}, cost={cost(w):.2f}")

def qolda(w,x,y):
  pred = (1/4)*np.sum((w*x-y)**2)
  return pred
for w in [2,3,4,5]:
  print(f" qolda w ={w}, cost={qolda(w,x,y):.2f}")


def farq(cost, qolda):
  pred=cost-qolda
  return pred
for w in [2,3,4,5]:
  print(f" qolda hisoblangan va simulyatsiyani w={w} va cost={cost(w):.2f} da farq ={farq(cost(w),qolda(w,x,y)):.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([3, 5, 4, 7], dtype=float)
y = np.array([5, 2, 4, 9], dtype=float)
N = len(x)

lr = 0.01
tol = 1e-5
iter = 1000

w_tarix = [0.0]

for i in range(iter):
    w_hozirgi = w_tarix[-1]
    dw = (2 / N) * np.sum((w_hozirgi * x - y) * x)

    w_yangi = w_hozirgi - lr * dw
    w_tarix.append(w_yangi)

    if len(w_tarix) > 1 and abs(w_tarix[-1] - w_tarix[-2]) < tol:
        print(f"To'xtash sharti bajarildi!")
        break

plt.figure(figsize=(8, 4))
plt.plot(w_tarix, 'o-', color='green', label='w qiymati')
plt.axhline(y=w_tarix[-1], color='r', linestyle='--', label=f' optimal w = {w_tarix[-1]:.4f}')
plt.title("w parametrining yaqinlashish grafigi")
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend()
plt.show()
```

```python
import numpy as np# b qo'shilganda w va b bo'yicha alohida gardient chiqar ekan.

x = np.array([3, 5, 4, 7])
y = np.array([5, 2, 4, 9])


def gradient(w, b):
  pred_y = w * x + b
  dw = np.mean(2 * x * (pred_y - y))
  db = np.mean(2 * (pred_y - y))
  return dw, db
for w, b in [(2, 0), (3, 1), (4, 0.5)]:
  dw, db = gradient(w, b)
  print(f"w={w}, b={b}   dw={dw:.2f}, db={db:.2f}")
```
